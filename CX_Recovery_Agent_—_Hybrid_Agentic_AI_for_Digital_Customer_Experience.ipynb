```python
def ai_classify_issue(message):
    message_lower = message.lower()

    # Hard safety guardrail
    safety_words = [
        "smoke",
        "smoking",
        "burning",
        "burnt",
        "spark",
        "sparking",
        "fire",
        "overheating"
    ]

    for word in safety_words:
        if word in message_lower:
            return "safety issue", 1.0

    # AI-based classification
    labels = [
        "safety issue",
        "installation issue",
        "technical malfunction",
        "documentation request",
        "usability issue",
        "customer service complaint",
        "general query"
    ]

    result = classifier(message, labels)

    issue_type = result["labels"][0]
    confidence = result["scores"][0]

    # Confidence guardrail
    if confidence < 0.60:
        return "uncertain", confidence

    return issue_type, confidence
```

```python
issue, confidence = ai_classify_issue("My charger started smoking.")

print("Issue:", issue)
print("Confidence:", confidence)
```
